```python

# Select one household to examine its readings
sample_household = df["household_ID"].dropna().iloc[0]

sample = df[df["household_ID"] == sample_household].copy()

# Convert date and time into a single timestamp
sample["timestamp"] = pd.to_datetime(
    sample["DATE"].astype(str) + " " + sample["TIME"].astype(str),
    errors="coerce"
)

# Convert meter readings to numeric values
sample["meter_reading"] = pd.to_numeric(
    sample["TOTAL_IMPORT (kWh)"],
    errors="coerce"
)

# Sort readings chronologically
sample = sample.dropna(
    subset=["timestamp", "meter_reading"]
).sort_values("timestamp")

# Calculate changes between consecutive readings
sample["time_gap_hours"] = (
    sample["timestamp"].diff().dt.total_seconds() / 3600
)

sample["consumption_change_kwh"] = (
    sample["meter_reading"].diff()
)

print("Sample household:", sample_household)
print("Number of valid readings:", len(sample))

display(sample[[
    "household_ID",
    "timestamp",
    "meter_reading",
    "time_gap_hours",
    "consumption_change_kwh"
]].head(15))
```

```python

# Prepare all household readings
df["timestamp"] = pd.to_datetime(
    df["DATE"].astype(str) + " " + df["TIME"].astype(str),
    errors="coerce"
)

df["meter_reading"] = pd.to_numeric(
    df["TOTAL_IMPORT (kWh)"],
    errors="coerce"
)

df = df.dropna(
    subset=["household_ID", "timestamp", "meter_reading"]
).copy()

df = df.sort_values(["household_ID", "timestamp"])

# Calculate changes within each household
df["time_gap_hours"] = (
    df.groupby("household_ID")["timestamp"]
    .diff()
    .dt.total_seconds() / 3600
)

df["consumption_change_kwh"] = (
    df.groupby("household_ID")["meter_reading"].diff()
)

# Summarize the readings
print("Total valid readings:", len(df))
print("Households:", df["household_ID"].nunique())

print("\nTime-gap summary (hours):")
print(df["time_gap_hours"].describe())

print("\nConsumption-change summary (kWh):")
print(df["consumption_change_kwh"].describe())

print("\nNegative consumption changes:",
      (df["consumption_change_kwh"] < 0).sum())

print("Zero consumption changes:",
      (df["consumption_change_kwh"] == 0).sum())

print("Gaps longer than 12 hours:",
      (df["time_gap_hours"] > 12).sum())
```

```python

# Inspect negative consumption changes
negative_changes = df[
    df["consumption_change_kwh"] < 0
]

print("Examples of negative meter changes:")
display(
    negative_changes[
        [
            "household_ID",
            "timestamp",
            "meter_reading",
            "time_gap_hours",
            "consumption_change_kwh"
        ]
    ].head(10)
)

# Inspect time gaps longer than 48 hours
long_gaps = df[
    df["time_gap_hours"] > 48
]

print("\nNumber of gaps longer than 48 hours:",
      len(long_gaps))

print("\nExamples of long gaps:")
display(
    long_gaps[
        [
            "household_ID",
            "timestamp",
            "meter_reading",
            "time_gap_hours",
            "consumption_change_kwh"
        ]
    ].head(10)
)
```

```python

# Step 12: Build an initial clean 6-hour dataset

clean_df = df.copy()

# Sort records by household and timestamp
clean_df = clean_df.sort_values(
    ["household_ID", "timestamp"]
).copy()

# Calculate time gaps and meter-reading differences
clean_df["time_gap_hours"] = (
    clean_df.groupby("household_ID")["timestamp"]
    .diff()
    .dt.total_seconds() / 3600
)

clean_df["consumption_change_kwh"] = (
    clean_df.groupby("household_ID")["meter_reading"]
    .diff()
)

# Keep approximately 6-hour intervals
clean_df = clean_df[
    clean_df["time_gap_hours"].between(4, 8)
].copy()

# Exclude negative consumption differences
clean_df = clean_df[
    clean_df["consumption_change_kwh"] >= 0
].copy()

# Calculate consumption per hour to account for small timing differences
clean_df["consumption_kwh_per_hour"] = (
    clean_df["consumption_change_kwh"]
    / clean_df["time_gap_hours"]
)

print("Initial clean dataset created!")
print("Rows:", len(clean_df))
print("Households:", clean_df["household_ID"].nunique())

print("\nTime-gap summary:")
print(clean_df["time_gap_hours"].describe())

print("\nConsumption difference summary:")
print(clean_df["consumption_change_kwh"].describe())

print("\nFirst 10 clean records:")
display(clean_df[[
    "household_ID",
    "timestamp",
    "time_gap_hours",
    "consumption_change_kwh",
    "consumption_kwh_per_hour"
]].head(10))
```

```python
# Step 20: Identify extreme jumps between consecutive readings

jump_check = df[
    ["household_ID", "timestamp", "meter_reading"]
].copy()

jump_check = (
    jump_check.sort_values(["household_ID", "timestamp"])
    .reset_index(drop=True)
)

jump_check["change_kwh"] = (
    jump_check.groupby("household_ID")["meter_reading"].diff()
)

# Large absolute jumps can indicate inconsistent readings
extreme_jumps = jump_check[
    jump_check["change_kwh"].abs() > 100
].copy()

print("Total extreme jumps:", len(extreme_jumps))
print(
    "Households affected:",
    extreme_jumps["household_ID"].nunique()
)

print("\nTop 15 extreme jumps:")
display(
    extreme_jumps.assign(
        absolute_change=extreme_jumps["change_kwh"].abs()
    )
    .nlargest(15, "absolute_change")[
        ["household_ID", "timestamp", "meter_reading", "change_kwh"]
    ]
)
```

```python
# Step 21: Examine the timing of extreme jumps

extreme_jumps = extreme_jumps.copy()

# Add time gaps using the chronologically sorted data
jump_timing = df[
    ["household_ID", "timestamp", "meter_reading"]
].copy()

jump_timing = (
    jump_timing.sort_values(["household_ID", "timestamp"])
    .reset_index(drop=True)
)

jump_timing["time_gap_hours"] = (
    jump_timing.groupby("household_ID")["timestamp"]
    .diff()
    .dt.total_seconds() / 3600
)

jump_timing["change_kwh"] = (
    jump_timing.groupby("household_ID")["meter_reading"].diff()
)

extreme_jumps = jump_timing[
    jump_timing["change_kwh"].abs() > 100
].copy()

print("Extreme jumps:", len(extreme_jumps))

print("\nTime gaps before extreme jumps:")
print(extreme_jumps["time_gap_hours"].describe())

print("\nExtreme jumps by time-gap category:")
print(
    pd.cut(
        extreme_jumps["time_gap_hours"],
        bins=[0, 0.1, 1, 4, 8, float("inf")],
        labels=["Under 6 min", "6 min–1 hr", "1–4 hr", "4–8 hr", "Over 8 hr"]
    ).value_counts().sort_index()
)
```

```python
# Step 23: Find households with the most extreme meter-reading jumps

extreme_jumps["absolute_change_kwh"] = (
    extreme_jumps["change_kwh"].abs()
)

household_jump_summary = (
    extreme_jumps.groupby("household_ID")
    .agg(
        extreme_jump_count=("change_kwh", "size"),
        maximum_jump_kwh=("absolute_change_kwh", "max"),
        median_time_gap_hours=("time_gap_hours", "median")
    )
    .sort_values("extreme_jump_count", ascending=False)
)

print("Top 10 households with extreme jumps:")
print(household_jump_summary.head(10).round(2))
```

```python
# Step 36: Split the dataset chronologically

model_df = model_df.sort_values("timestamp").copy()

unique_dates = sorted(model_df["date"].unique())

split_position = int(len(unique_dates) * 0.8)
split_date = unique_dates[split_position]

train_df = model_df[model_df["date"] < split_date].copy()
test_df = model_df[model_df["date"] >= split_date].copy()

print("Training period:")
print(train_df["timestamp"].min(), "to", train_df["timestamp"].max())

print("\nTesting period:")
print(test_df["timestamp"].min(), "to", test_df["timestamp"].max())

print("\nTraining rows:", len(train_df))
print("Testing rows:", len(test_df))

print("\nTraining percentage:",
      round(len(train_df) / len(model_df) * 100, 2), "%")
print("Testing percentage:",
      round(len(test_df) / len(model_df) * 100, 2), "%")
```

```python
# Step 37: Create the next-interval prediction target

model_df = model_df.sort_values(
    ["household_ID", "timestamp"]
).copy()

model_df["hour"] = model_df["timestamp"].dt.hour
model_df["day_of_week"] = model_df["timestamp"].dt.dayofweek
model_df["day_of_year"] = model_df["timestamp"].dt.dayofyear

# Target: next recorded consumption rate for the same household
model_df["target_next_rate"] = (
    model_df.groupby("household_ID")["consumption_kwh_per_hour"]
    .shift(-1)
)

# Remove rows where the next observation is unavailable
prediction_df = model_df.dropna(
    subset=["target_next_rate"]
).copy()

print("Rows available for prediction:", len(prediction_df))
print("\nInput features:")
print(["hour", "day_of_week", "day_of_year"])

print("\nTarget:")
print("target_next_rate")

print("\nSample rows:")
print(
    prediction_df[
        [
            "household_ID",
            "timestamp",
            "hour",
            "day_of_week",
            "day_of_year",
            "target_next_rate"
        ]
    ].head(10).to_string(index=False)
)
```

```python
# Step 38: Verify that the target is genuinely in the future

prediction_df["next_timestamp"] = (
    prediction_df.groupby("household_ID")["timestamp"].shift(-1)
)

prediction_df["next_gap_hours"] = (
    prediction_df["next_timestamp"] - prediction_df["timestamp"]
).dt.total_seconds() / 3600

# Rows where a target exists should have a next timestamp
valid_target_rows = prediction_df.dropna(
    subset=["target_next_rate", "next_timestamp"]
)

print("Rows with valid next targets:", len(valid_target_rows))

print("\nGap to next observation (hours):")
print(valid_target_rows["next_gap_hours"].describe().round(3))

print(
    "\nTargets with next observation within 4–8 hours:",
    valid_target_rows["next_gap_hours"].between(4, 8).sum()
)

print(
    "Targets with next observation outside 4–8 hours:",
    (~valid_target_rows["next_gap_hours"].between(4, 8)).sum()
)
```

```python
# Step 39: Correctly align the future target and timestamp

model_df = model_df.sort_values(
    ["household_ID", "timestamp"]
).copy()

# Find the next observation for the same household
model_df["target_next_rate"] = (
    model_df.groupby("household_ID")["consumption_kwh_per_hour"].shift(-1)
)

model_df["target_next_timestamp"] = (
    model_df.groupby("household_ID")["timestamp"].shift(-1)
)

# Calculate the actual time gap to that target
model_df["target_gap_hours"] = (
    model_df["target_next_timestamp"] - model_df["timestamp"]
).dt.total_seconds() / 3600

# Keep rows where both the future target and timestamp exist
prediction_df = model_df.dropna(
    subset=["target_next_rate", "target_next_timestamp"]
).copy()

print("Rows with valid future targets:", len(prediction_df))

print("\nGap to future target (hours):")
print(prediction_df["target_gap_hours"].describe().round(3))

# Keep only targets 4–8 hours into the future
prediction_df = prediction_df[
    prediction_df["target_gap_hours"].between(4, 8)
].copy()

print("\nRows after keeping 4–8 hour targets:", len(prediction_df))
print(
    "Unique households:",
    prediction_df["household_ID"].nunique()
)
print(
    "Target gap range:",
    prediction_df["target_gap_hours"].min(),
    "to",
    prediction_df["target_gap_hours"].max(),
    "hours"
)
```

```python
# Step 40: Inspect the prediction dataset

print("Dataset shape:", prediction_df.shape)

print("\nAvailable columns:")
print(prediction_df.columns.tolist())

print("\nMissing values in important columns:")
important_columns = [
    "household_ID",
    "timestamp",
    "consumption_kwh_per_hour",
    "target_next_rate",
    "target_gap_hours"
]

print(prediction_df[important_columns].isnull().sum())

print("\nTarget variable summary:")
print(prediction_df["target_next_rate"].describe().round(4))

print("\nSample records:")
display(prediction_df[important_columns].head(10))
```

```python
# Step 41: Chronological train-test split

# Sort by timestamp
prediction_df = prediction_df.sort_values("timestamp").copy()

# Use the first 80% of unique dates for training
unique_dates = sorted(prediction_df["timestamp"].dt.date.unique())

split_position = int(len(unique_dates) * 0.8)
split_date = unique_dates[split_position]

train_df = prediction_df[
    prediction_df["timestamp"].dt.date < split_date
].copy()

test_df = prediction_df[
    prediction_df["timestamp"].dt.date >= split_date
].copy()

print("Split date:", split_date)

print("\nTraining dataset")
print("Rows:", len(train_df))
print("Start:", train_df["timestamp"].min())
print("End:", train_df["timestamp"].max())

print("\nTesting dataset")
print("Rows:", len(test_df))
print("Start:", test_df["timestamp"].min())
print("End:", test_df["timestamp"].max())

print("\nTraining percentage:", round(len(train_df) / len(prediction_df) * 100, 2), "%")
print("Testing percentage:", round(len(test_df) / len(prediction_df) * 100, 2), "%")
```

```python
# Step 42: Select features and target

feature_columns = [
    "consumption_kwh_per_hour",
    "hour",
    "day_of_week",
    "day_of_year",
    "target_gap_hours"
]

target_column = "target_next_rate"

X_train = train_df[feature_columns].copy()
y_train = train_df[target_column].copy()

X_test = test_df[feature_columns].copy()
y_test = test_df[target_column].copy()

print("Input features:", feature_columns)

print("\nX_train shape:", X_train.shape)
print("y_train shape:", y_train.shape)

print("\nX_test shape:", X_test.shape)
print("y_test shape:", y_test.shape)

print("\nMissing values in X_train:", X_train.isnull().sum().sum())
print("Missing values in X_test:", X_test.isnull().sum().sum())

print("\nTraining feature sample:")
display(X_train.head())

print("\nTraining target sample:")
display(y_train.head())
```

```python
# Step 43: Train a baseline Linear Regression model

from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Use only features available at the current observation
baseline_features = [
    "consumption_kwh_per_hour",
    "hour",
    "day_of_week",
    "day_of_year"
]

X_train_base = train_df[baseline_features]
X_test_base = test_df[baseline_features]

y_train = train_df["target_next_rate"]
y_test = test_df["target_next_rate"]

# Train model
lr_model = LinearRegression()
lr_model.fit(X_train_base, y_train)

# Make predictions
lr_predictions = lr_model.predict(X_test_base)

# Evaluate model
mae = mean_absolute_error(y_test, lr_predictions)
rmse = np.sqrt(mean_squared_error(y_test, lr_predictions))
r2 = r2_score(y_test, lr_predictions)

print("Linear Regression Baseline Results")
print("----------------------------------")
print("MAE:", round(mae, 4))
print("RMSE:", round(rmse, 4))
print("R² Score:", round(r2, 4))
```

```python
# Step 45: Train a Neural Network model

from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.neural_network import MLPRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Use the same features as Linear Regression
nn_features = [
    "consumption_kwh_per_hour",
    "hour",
    "day_of_week",
    "day_of_year"
]

X_train_nn = train_df[nn_features]
X_test_nn = test_df[nn_features]

y_train_nn = train_df["target_next_rate"]
y_test_nn = test_df["target_next_rate"]

# Scaling + Multilayer Neural Network
nn_model = Pipeline([
    ("scaler", StandardScaler()),
    ("mlp", MLPRegressor(
        hidden_layer_sizes=(64, 32),
        activation="relu",
        solver="adam",
        learning_rate_init=0.001,
        max_iter=100,
        early_stopping=True,
        random_state=42
    ))
])

# Train
nn_model.fit(X_train_nn, y_train_nn)

# Predict
nn_predictions = nn_model.predict(X_test_nn)

# Evaluate
nn_mae = mean_absolute_error(y_test_nn, nn_predictions)
nn_rmse = np.sqrt(mean_squared_error(y_test_nn, nn_predictions))
nn_r2 = r2_score(y_test_nn, nn_predictions)

print("Neural Network Results")
print("----------------------")
print("MAE:", round(nn_mae, 4))
print("RMSE:", round(nn_rmse, 4))
print("R² Score:", round(nn_r2, 4))
```

```python
# Step 47: Compare model performance

import pandas as pd

results_df = pd.DataFrame({
    "Model": ["Linear Regression", "Neural Network (MLP)"],
    "MAE": [mae, nn_mae],
    "RMSE": [rmse, nn_rmse],
    "R2 Score": [r2, nn_r2]
})

print("Model Performance Comparison")
display(results_df.round(4))

# Identify the best model for each metric
print("\nBest model by MAE:", results_df.loc[results_df["MAE"].idxmin(), "Model"])
print("Best model by RMSE:", results_df.loc[results_df["RMSE"].idxmin(), "Model"])
print("Best model by R²:", results_df.loc[results_df["R2 Score"].idxmax(), "Model"])
```

```python
# Step 53: Linear Regression with log-transformed target

from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Use the same input features
features = [
    "consumption_kwh_per_hour",
    "hour",
    "day_of_week",
    "day_of_year"
]

X_train_log = train_df[features]
X_test_log = test_df[features]

y_train_log = np.log1p(train_df["target_next_rate"])
y_test_original = test_df["target_next_rate"].to_numpy()

# Train on the log-transformed target
log_lr_model = Pipeline([
    ("scaler", StandardScaler()),
    ("regression", LinearRegression())
])

log_lr_model.fit(X_train_log, y_train_log)

# Predict in log scale, then convert back
log_predictions = log_lr_model.predict(X_test_log)
log_predictions_original = np.expm1(log_predictions)

# Consumption rate cannot be negative
log_predictions_original = np.maximum(log_predictions_original, 0)

# Evaluate on the original scale
log_mae = mean_absolute_error(
    y_test_original, log_predictions_original
)
log_rmse = np.sqrt(mean_squared_error(
    y_test_original, log_predictions_original
))
log_r2 = r2_score(
    y_test_original, log_predictions_original
)

print("Log-Target Linear Regression Results")
print("------------------------------------")
print("MAE:", round(log_mae, 4))
print("RMSE:", round(log_rmse, 4))
print("R² Score:", round(log_r2, 4))
```
